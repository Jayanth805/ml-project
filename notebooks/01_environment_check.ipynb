## Verify GPU Availability for Training

```python
# Comprehensive GPU check
def check_gpu():
    """Check GPU availability and print diagnostic info."""
    if not torch.cuda.is_available():
        print("="*50)
        print("GPU NOT AVAILABLE")
        print("="*50)
        print("Training will run on CPU (slower).")
        print("For GPU training, use Google Colab with GPU runtime.")
        return False
    
    print("="*50)
    print("GPU AVAILABLE")
    print("="*50)
    print(f"GPU Name: {torch.cuda.get_device_name(0)}")
    
    props = torch.cuda.get_device_properties(0)
    print(f"Total Memory: {props.total_memory / 1e9:.2f} GB")
    print(f"Compute Capability: {props.major}.{props.minor}")
    
    # Test GPU tensor operations
    try:
        x = torch.randn(1000, 1000).cuda()
        y = torch.matmul(x, x)
        print("✓ GPU tensor operations working")
        del x, y
        torch.cuda.empty_cache()
    except Exception as e:
        print(f"✗ GPU tensor operations failed: {e}")
        return False
    
    return True

gpu_available = check_gpu()
```
